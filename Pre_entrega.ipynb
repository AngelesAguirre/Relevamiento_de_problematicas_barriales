## **1. Creación de la lista de reglamos**

La primera etapa va a consistir en la creación de una lista que, ahora vacía, va a conservar los reclamos que se vayan cargando en el sistema.

En esta etapa se crean las categorías y subcategorías que van a permitir generar diccionarios.

---
**NOTA**
Una vez que ejecute esta celda de codigo NO LA VUELVO A EJECUTAR porque la lsita se va a reiniciar

```python
# 1. Lista principal,vacía, donde se van a guardar todos los reclamos
Lista_Reclamos = []

# 2. Diccionario con las categorías y subcategorías posibles del sistema
  # No quise dejar esto para autocompletar libremente por el usuario porque posteriormente va a producir un problema de categorizacion.
  # Al Agrupar lo más especificamente las problematicas, puede pensarse en un posterior mapeo

# Categoria 1
Categorias = {"1": { "Nombre": "Alumbrado y señalización",
                    "Subcategorias": ["Luminaria quemada",
                                      "Semáforo vehicular",
                                      "Semáforo peatonal",
                                      "Cables caídos"]},
# Categoria 2
              "2": {"Nombre": "Veredas y accesibilidad",
                    "Subcategorias": ["Rampa rota",
                                      "Rampa obstruida",
                                      "Cordón roto",
                                      "Vereda rota",
                                      "Vereda obstruida"]},
# Categoria 3
              "3": {"Nombre": "Infraestructura vial",
                    "Subcategorias": ["Bache",
                                      "Calle anegada", # acumulación de agua
                                      "Obstrucción vial",
                                      "Carril reducido",
                                      "Calzada deteriorada"]},
# Categoria 4
              "4": {"Nombre": "Higiene urbana",
                    "Subcategorias": ["Basural",
                                      "Residuos acumulados",
                                      "Contenedor desbordado"]}}

# 3. Mensaje inicial para verificar que el sistema comenzó correctamente
print("Sistema de Registro de Problemáticas Barriales iniciado correctamente. \nPrototipo en curso \n")
```

## **2. Creación de la función de carga de reclamos**

En esta etapa se desarrolla la función (reutilizable con def) principal donde se van a registrar nuevos reclamos dentro del sistema.

La funcion va a permitir mostrar categorias y subcategorias; validar opciones ingresadas; solicitar datos territoriales y almacenar cada reclamo dentro de la lista principal.

La utilización de una función permite mantener el código más organizado, reutilizable y fácil de mantener a medida que el proyecto aumente su complejidad.

```python
# 1. Función para agregar un nuevo reclamo al sistema
def Agregar_Reclamo():
    print("\nCARGA DE REGLAMOS\n")

# 2. Dentro de la función se muestran las categorías disponibles
    print("\nCategorías disponibles:")

# 3. Con el for vamos a mostrar una a una las categorias con su número (clave)
    for clave in Categorias:
        print(clave + ".", Categorias[clave]["Nombre"])

#4. Se pide al usuario elegir una categoría
# Se eliminan los espacion en blanco al inicio y al final con el strip
    Opcion_Categoria = input("\nSeleccione el número de la categoría: ").strip()

# Se fuerza a que solo puedan elegir entre las categorias existentes, sino que de error
    if Opcion_Categoria in Categorias:
        print("\nCategoría válida.")

# Si bien podria corroborarse tambien con isdigit y colocando el rango valido, el diciconario ya fucniona por si solo como validación

# 5. Guardo la categoria elegida en una variable para poder mostrar las opciones correspondientes
        Categoria_Elegida = Categorias[Opcion_Categoria]

# También guardo la lista de subcategorías disponibles para trabajar más cómodo
        Subcategorias_Disponibles = Categoria_Elegida["Subcategorias"]

# 6. La imprimo solo para que se corrobore de que categoria vamos a seleccionar las subcategorias
        print("\nSubcategorías disponibles para la categoría " + Categoria_Elegida["Nombre"] + ": \n")

# 7. Se muestran las subcategorias usando su posicion en la lista
# Con el len() le doy valores desde 0 a las subcategorias y el for recorre en indice
        for i in range(len(Subcategorias_Disponibles)):
            print(str(i + 1) + ".",
                  Subcategorias_Disponibles[i])

# 8. Ahora se le pide llenar la opcion y quito espacios en blanco
        Opcion_Subcategoria = input("\nSeleccione el número de la subcategoría: \n").strip()

# 9. Se valida que la opción ingresada sea un número
        if Opcion_Subcategoria.isdigit():
            Opcion_Subcategoria = int(Opcion_Subcategoria)

# Se valida que el número esté dentro del rango de opciones
            if Opcion_Subcategoria >= 1 and Opcion_Subcategoria <= len(Subcategorias_Disponibles):
                Subcategoria_Elegida = Subcategorias_Disponibles[Opcion_Subcategoria - 1]

                print("\nSubcategoría seleccionada correctamente.")

# Se piden los datos territoriales
                Provincia = input("\nIngrese la provincia: ").strip().title()
                Municipio = input("Ingrese el municipio o partido: ").strip().title()
                Calle = input("Ingrese la calle: ").strip().title()
                Altura = input("Ingrese la altura: ").strip()

# Valido que los campos territoriales no estén vacíos
                if Provincia == "" or Municipio == "" or Calle == "" or Altura == "":
                    print("\nERROR: No se pueden dejar campos vacíos. \n")

                else:
# Si todo esta completo se crea el diccionario del reclamo
                    Reclamo = {"Categoria": Categoria_Elegida["Nombre"],
                               "Subcategoria": Subcategoria_Elegida,
                               "Provincia": Provincia,
                               "Municipio": Municipio,
                               "Calle": Calle,
                               "Altura": Altura}

# Agrego el reclamo a la lista principal
                    Lista_Reclamos.append(Reclamo)

                    print("\nReclamo agregado correctamente.")

# Si no selecciono un numero que este dentro de las subcategorias se imprime un error
            else:
                print("\nERROR: Debe seleccionar una subcategoría válida. \n")

# Si no se ingresan numeros tambien se da error
        else:
            print("\nERROR: Debe ingresar un número.\n")
# SE fuerza a seleccionar una categoria valida
    else:
        print("\nERROR: Debe seleccionar una categoría válida.")
```

## **3. Carga de reclamos en el sistema**

Esta etapa, al ejecutar la funcion, se permite el ingreso de reclamos urbanos mediante la interacción con el usuario, solicitando los datos del bloque anterior. Permitiendo seleccionar una categoría y una subcategoría; ingresar datos de ubicación territorial y almacenar la información dentro de Lista_Reclamos.

Cada reclamo será guardado como un diccionario independiente dentro de la lista, permitiendo organizar y acceder a la información de manera clara y estructurada; y con el bucle permitimos que la persona cargue más de un reclamo por sesión

---
**NOTA**

Cada vez que ejecute esta celda de codigo voy a poder agregar reclamos a la lista sin eliminar los existentes, sin reiniciarla

```python
# Variable para mantener activo el sistema, permite que el codigo siga ejecutandose
Sistema_Activo = True

while Sistema_Activo:

    print("\n MENÚ PRINCIPAL ")
    print("1. Agregar reclamo")
    print("2. Salir")

# Se solicita seleccionar una opción
    Opcion_Menu = input("\nSeleccione una opción: ").strip()

# Opción 1: agregar un nuevo reclamo
    if Opcion_Menu == "1":
        Agregar_Reclamo()

# Opción 2: finalizar el programa
    elif Opcion_Menu == "2":
        print("\nSistema finalizado correctamente.")
        Sistema_Activo = False

# Si la opción no existe, se muestra error
    else:
        print("\nERROR: Debe seleccionar una opción válida.")
```

## **4. Incorporación de funciones de visualización y gestión de reclamos**

Luego cargar los datos, el siguiente paso es desarrollar herramientas que permitan administrar la información almacenada dentro de Lista_Reclamos.

La idea es transformar el proyecto en un sistema más dinámico y funcional, permitiendo registrar información, consultarla y administrarla posteriormente.

```python
# 1. Función para mostrar todos los reclamos cargados en el sistema
def Mostrar_Reclamos():
    print("\nLISTA DE RECLAMOS REGISTRADOS\n")

    # Se valida si la lista esta vacía
    if len(Lista_Reclamos) == 0:
        print("No hay reclamos cargados en el sistema.")

    # Pero si no esta vacia se recorre la lista con for
    else:
        for i in range(len(Lista_Reclamos)):

            Reclamo = Lista_Reclamos[i]

            print("Reclamo", i + 1)
            print("Categoría:", Reclamo["Categoria"])
            print("Subcategoría:", Reclamo["Subcategoria"])
            print("Provincia:", Reclamo["Provincia"])
            print("Municipio:", Reclamo["Municipio"])
            print("Calle:", Reclamo["Calle"])
            print("Altura:", Reclamo["Altura"])

# 2. Se llama a la función
Mostrar_Reclamos()

# 3. Se da el mensaje de finalizacion
print("La lista de reclamos a sido presentada completamente")
```

## **5. Búsqueda y eliminación de reclamos**

En esta etapa se incorporan dos nuevas funciones necesarias para completar las funcionalidades principales del sistema.

Por un lado, se desarrolla una función de búsqueda que permite localizar reclamos cargados previamente dentro de Lista_Reclamos.
Por otro lado, se incorpora una función de eliminación que permite borrar un reclamo específico a partir de su número de registro. Para evitar errores, primero se mostrarán los reclamos disponibles y luego se validará que la opción ingresada corresponda a un registro existente.

```python
# 1. Función para buscar reclamos dentro del sistema
def Buscar_Reclamo():

    print("\nBUSQUEDA DE RECLAMOS\n")

# 2. Se valida si la lista está vacía
    if len(Lista_Reclamos) == 0:
        print("No hay reclamos cargados para buscar.")

# 3. Se solicita una palabra de búsqueda
    else:
        Busqueda = input("Ingrese una palabra para buscar: ").strip().lower()

# 4. Variable para controlar si se encontraron coincidencias
        Encontrado = False

# 5. Se recorren los reclamos cargados
        for i in range(len(Lista_Reclamos)):
            Reclamo = Lista_Reclamos[i]

# 5. Se busca la palabra en distintos campos del reclamo
            if (Busqueda in Reclamo["Categoria"].lower() or
                Busqueda in Reclamo["Subcategoria"].lower() or
                Busqueda in Reclamo["Provincia"].lower() or
                Busqueda in Reclamo["Municipio"].lower() or
                Busqueda in Reclamo["Calle"].lower() or
                Busqueda in Reclamo["Altura"].lower()):

                print("\nReclamo encontrado:", i + 1)
                print("Categoría:", Reclamo["Categoria"])
                print("Subcategoría:", Reclamo["Subcategoria"])
                print("Provincia:", Reclamo["Provincia"])
                print("Municipio:", Reclamo["Municipio"])
                print("Calle:", Reclamo["Calle"])
                print("Altura:", Reclamo["Altura"])

                Encontrado = True

# 6. Si no hubo coincidencias, se informa al usuario
        if Encontrado == False:
            print("\nNo se encontraron reclamos con esa búsqueda.")

# 7. Se llama a la funcion
Buscar_Reclamo()
```

```python
# 1. Función para eliminar un reclamo del sistema
def Eliminar_Reclamo():
    print("\nELIMINACION DE RECLAMOS\n")

# 2. Se valida si la lista está vacía
    if len(Lista_Reclamos) == 0:
        print("No hay reclamos cargados para eliminar.")

# 3. Primero se muestran los reclamos cargados
    else:
        Mostrar_Reclamos()

# 4. Se solicita el número del reclamo a eliminar
        Opcion_Eliminar = input("\nIngrese el número del reclamo que desea eliminar: ").strip()

# 5. Se valida que se haya ingresado un número y que sea entero
        if Opcion_Eliminar.isdigit():
            Opcion_Eliminar = int(Opcion_Eliminar)

# 6. Se valida que el número corresponda a un reclamo existente
            if Opcion_Eliminar >= 1 and Opcion_Eliminar <= len(Lista_Reclamos):
                Reclamo_Eliminado = Lista_Reclamos.pop(Opcion_Eliminar - 1)

# En tal caso se confirma que se elimino
                print("\nReclamo eliminado correctamente:")
                print(Reclamo_Eliminado)

# Sino se da error y se pide que se ingrese un numero valido
            else:
                print("\nERROR: El número ingresado no corresponde a un reclamo existente.")

        else:
            print("\nERROR: Debe ingresar un número válido.")


# 7. Llamo a la funcion
Eliminar_Reclamo()
```
